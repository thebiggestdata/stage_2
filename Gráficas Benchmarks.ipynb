```python
import pandas as pd
import matplotlib.pyplot as plt

# Leer el CSV
df = pd.read_csv("jmh-stopword.csv")

# Asegurar tipo numérico
df["Param: textLength"] = df["Param: textLength"].astype(int)
df["Score"] = df["Score"].astype(float)

# Crear la gráfica
plt.figure(figsize=(8,5))
plt.plot(df["Param: textLength"], df["Score"], marker='o', linewidth=2, color='seagreen')

# Etiquetas y título
plt.title("Benchmark Indexing-StopWordFilter - Tiempo promedio por tamaño de texto", fontsize=12, weight='bold')
plt.xlabel("Longitud del texto (caracteres)")
plt.ylabel("Tiempo medio (ms/op)")
plt.grid(True, linestyle="--", alpha=0.6)
plt.tight_layout()

# Mostrar
plt.show()
```
